```python
import pandas as pd

print("="*50)
print("FINAL RESEARCH EVALUATION REPORT")
print("="*50)

csv_baseline_small = "../data/benchmark/eval_01_baseline_small.csv"
csv_formula_small = "../data/benchmark/eval_02_formula_small.csv"
csv_baseline_large = "../data/benchmark/eval_03_baseline_large.csv"
csv_formula_large = "../data/benchmark/eval_04_formula_large.csv"

def calculate_accuracy(file_path, name):
    try:
        df = pd.read_csv(file_path)
        correct = (df['Expected_Answer'].astype(str) == df['Predicted_Answer'].astype(str)).sum()
        total = len(df)
        acc = (correct / total) * 100
        print(f"-> {name}: {acc:.2f}% ({correct}/{total})")
        return df
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
        return None

print("\n[A] SMALL SCALE RESULTS (1 PDF, 7 Questions)")
df_b_small = calculate_accuracy(csv_baseline_small, "Baseline Method")
df_f_small = calculate_accuracy(csv_formula_small, "Formula-Aware Method")

print("\n[B] MASSIVE SCALE RESULTS (3 PDFs, 50 Questions)")
df_b_large = calculate_accuracy(csv_baseline_large, "Baseline Method")
df_f_large = calculate_accuracy(csv_formula_large, "Formula-Aware Method")

print("\n[C] GENERATING FINAL COMPARISON EXCEL...")
if df_f_large is not None:
    df_f_large['Status'] = df_f_large.apply(
        lambda row: "BENAR" if str(row['Expected_Answer']) == str(row['Predicted_Answer']) else "SALAH", 
        axis=1
    )
    
    output_path = "../data/benchmark/FINAL_Formula_Large_Analysis.csv"
    final_df = df_f_large[['Question', 'Expected_Answer', 'Predicted_Answer', 'Status', 'Retrieved_Context']]
    final_df.to_csv(output_path, index=False)
    print(f"-> Question analysis details successfully saved in: {output_path}")

print("="*50)
print("EXPERIMENT COMPLETION: 100%")
print("="*50)
```
